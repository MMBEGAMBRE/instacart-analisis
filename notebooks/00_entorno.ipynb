# Taller: preparar el entorno de Instacart

Este cuaderno acompaña la preparación del repositorio `instacart-analisis` desde cero. Los comandos se muestran para Windows PowerShell; donde corresponde se incluye la alternativa para macOS/Linux. Ejecuta los comandos de terminal desde la ubicación indicada y las celdas de código en orden.

## 1. Verificar herramientas base

Se necesita Python 3.12 o superior y Git 2.x. Desde una terminal:

```powershell
python --version
git --version
```

Si `python` apunta a Python 2, usa `python3` en los comandos Python. Si Git no tiene identidad configurada, configúrala una sola vez con tu nombre y correo:

```powershell
git config --global user.name "Tu Nombre"
git config --global user.email "tu@correo.com"
```

## 2. Inicializar repositorio y estructura mínima

Desde la carpeta que contendrá el proyecto:

```powershell
mkdir instacart-analisis
cd instacart-analisis
git init -b main
mkdir data
New-Item data/.gitkeep -ItemType File
```

`data/.gitkeep` permite que Git conserve la carpeta vacía. Comprueba el estado inicial con `git status --short`.

## 3. Configurar `.gitignore` antes del primer commit

Crea el archivo en la raíz antes de ejecutar `git add`. En PowerShell:

```powershell
@(".venv/", "__pycache__/", ".ipynb_checkpoints/", "data/*.csv") | Set-Content -Encoding ascii .gitignore
```

El contenido debe ser:

```text
.venv/
__pycache__/
.ipynb_checkpoints/
data/*.csv
```

Así no se versionan el entorno, la caché, los checkpoints ni los CSV. Valida la exclusión del entorno con:

```powershell
git check-ignore -v .venv
```

## 4. Crear y activar `.venv`

Crea el entorno virtual y actívalo en PowerShell:

```powershell
python -m venv .venv
.\.venv\Scripts\Activate.ps1
python -c "import sys; print(sys.prefix)"
```

La ruta impresa debe terminar en `instacart-analisis\.venv`. En macOS/Linux, la activación es `source .venv/bin/activate`. Si PowerShell bloquea la activación, puedes permitir scripts locales para tu usuario con `Set-ExecutionPolicy -Scope CurrentUser RemoteSigned` y volver a activarlo.

## 5. Instalar pandas y generar `requirements.txt` (`v0.2`)

Con `.venv` activo:

```powershell
python -m pip install --upgrade pip
python -m pip install pandas
python -m pip freeze | Out-File -Encoding ascii requirements.txt
python -c "import pandas as pd, numpy as np; print(pd.__version__, np.__version__)"
python -c "import pandas as pd; print(pd.DataFrame({'aisle_id': [1, 2], 'aisle': ['soups', 'pasta']}).shape)"
```

`pip freeze` registra las versiones exactas, incluyendo NumPy, que pandas instala como dependencia. En PowerShell, `Out-File -Encoding ascii` conserva un formato de texto que `pip` puede leer en distintos sistemas.

## 6. Instalar JupyterLab en el mismo entorno (`v0.3`)

```powershell
python -m pip install jupyterlab
python -m pip freeze | Out-File -Encoding ascii requirements.txt
jupyter --version
jupyter lab
```

El navegador abrirá JupyterLab. Para detener el servidor, vuelve a la terminal y presiona `Ctrl + C`. Instalar JupyterLab dentro de `.venv` permite que el kernel encuentre pandas.

## 7. Crear `notebooks/00_entorno.ipynb` y validar pandas

Desde la raíz del proyecto crea `notebooks/`, abre JupyterLab y guarda este cuaderno como `notebooks/00_entorno.ipynb`. Selecciona el kernel Python de `.venv`. La primera celda de código, debajo, imprime las versiones de Python y pandas; ejecútala antes de continuar. Python debe ser 3.12 o superior y pandas debe importarse sin error.

## 8. Implementar y probar `franja`

La segunda celda define la función. `if` identifica la mañana, `elif` la tarde y `else` la noche. La prueba recorre las horas `8`, `13` y `21`.

## 9. Versionar ramas, commits y tags `v0.1` a `v0.4`

Guarda cada hito desde la raíz del repositorio. Usa las ramas de trabajo y merges indicados para conservar el historial de integración.

**v0.1: estructura inicial en `main`**

```powershell
git add .gitignore data/.gitkeep
git commit -m "chore: estructura inicial del proyecto"
git tag -a v0.1 -m "v0.1 - estructura, gitignore y entorno virtual"
git switch -c develop
```

**v0.2: pandas en `feat/entorno`**

```powershell
git switch -c feat/entorno
git add requirements.txt
git commit -m "feat: instalar pandas y registrar dependencias"
git tag -a v0.2 -m "v0.2 - pandas instalado y requirements.txt"
```

**v0.3: JupyterLab e integración de `feat/entorno`**

```powershell
git add requirements.txt
git commit -m "feat: instalar jupyterlab en el entorno del proyecto"
git tag -a v0.3 -m "v0.3 - jupyterlab en el entorno"
git switch develop
git merge --no-ff feat/entorno -m "chore: integrar feat/entorno en develop"
git branch -d feat/entorno
```

**v0.4: cuaderno e integración de `feat/cuaderno`**

```powershell
git switch -c feat/cuaderno
```

Guarda y ejecuta las dos celdas de este cuaderno, luego:

```powershell
git add notebooks/00_entorno.ipynb
git commit -m "feat: primer cuaderno con comprobación de pandas y función franja"
git tag -a v0.4 -m "v0.4 - cuaderno 00_entorno con pandas y franja"
git switch develop
git merge --no-ff feat/cuaderno -m "chore: integrar feat/cuaderno en develop"
git branch -d feat/cuaderno
```

Al terminar, `git tag` lista los cuatro hitos y `git status --short` permite comprobar si quedan cambios sin guardar.

````python
import sys
import pandas as pd

print(sys.version_info.major, sys.version_info.minor)
print(pd.__version__)
````

````python
def franja(order_hour_of_day):
    """Clasifica la hora del pedido en mañana, tarde o noche."""
    if order_hour_of_day < 12:
        return "mañana"
    elif order_hour_of_day < 19:
        return "tarde"
    else:
        return "noche"


for hora in [8, 13, 21]:
    print(hora, "->", franja(hora))
````
